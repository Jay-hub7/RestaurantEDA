```python
plt.figure(figsize=(8,5))

sns.histplot(df["Revenue"], kde=True)

plt.title("Distribution of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Profit"], kde=True)

plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Customers"], kde=True)

plt.title("Distribution of Customers")
plt.xlabel("Number of Customers")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Marketing_Spend"], kde=True)

plt.title("Distribution of Marketing Spend")
plt.xlabel("Marketing Spend")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Staff_Count"], kde=True)

plt.title("Distribution of Staff Count")
plt.xlabel("Staff Count")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Avg_Delivery_Min"], kde=True)

plt.title("Distribution of Average Delivery Time")
plt.xlabel("Average Delivery Time (Minutes)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Customer_Rating"], kde=True)

plt.title("Distribution of Customers Rating")
plt.xlabel("Customers Rating")
plt.ylabel("Frequency")

plt.show()
```

```python
branch_revenue = df.groupby("Branch")["Revenue"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=branch_revenue.index, y=branch_revenue.values)

plt.title("Average Revenue by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Revenue")
plt.xticks(rotation=45)

plt.show()
```

```python
branch_profit = df.groupby("Branch")["Profit"].mean().sort_values(ascending=False)

plt.figure(figsize=(10,5))
sns.barplot(x=branch_profit.index, y=branch_profit.values)

plt.title("Average Profit by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Profit")
plt.xticks(rotation=45)

plt.show()
```

```python
region_revenue = df.groupby("Region")["Revenue"].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=region_revenue.index, y=region_revenue.values)

plt.title("Average Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Average Revenue")
plt.xticks(rotation=45)

plt.show()
```

```python
store_performance = df.groupby("Store_Type")[["Revenue", "Profit"]].mean()

store_performance.plot(kind="bar", figsize=(9, 5))

plt.title("Average Revenue and Profit by Store Type")
plt.xlabel("Store Type")
plt.ylabel("Average Amount")
plt.xticks(rotation=0)
plt.legend(title="Metrics")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Marketing_Spend",
    y="Revenue",
    alpha=0.9
)

plt.title("Marketing Spend vs Revenue")
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Staff_Count",
    y="Customers",
    alpha=0.8
)

plt.title("Staff Count vs Customers")
plt.xlabel("Staff Count")
plt.ylabel("Customers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Avg_Delivery_Min",
    y="Customer_Rating",
    alpha=0.8
)

plt.title("Average Delivery Time vs Customer Rating")
plt.xlabel("Average Delivery Time (Minutes)")
plt.ylabel("Customer Rating")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Variables")
plt.show()
```

```python
print("Highest Average Revenue Branch:")
print(df.groupby("Branch")["Revenue"].mean().sort_values(ascending=False).head(1))

print("\nHighest Average Profit Branch:")
print(df.groupby("Branch")["Profit"].mean().sort_values(ascending=False).head(1))

print("\nHighest Average Revenue Region:")
print(df.groupby("Region")["Revenue"].mean().sort_values(ascending=False).head(1))

print("\nStore Type Performance:")
print(df.groupby("Store_Type")[["Revenue", "Profit"]].mean().round(2))

print("\nKey Correlations:")
print(df[[
    "Marketing_Spend", "Revenue", "Staff_Count",
    "Customers", "Avg_Delivery_Min", "Customer_Rating"
]].corr().round(2))
```
